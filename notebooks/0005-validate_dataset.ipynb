```python
import pandas as pd
import os 
from torch.utils.data import DataLoader
import warnings
import sys
sys.path.append('..')

from src.datasets import NIHChestXrayDataset
```

```python
data_dir = "/mnt/n4i_ds423/Employee_Folder/Jalil_Ahmed/NIH_Chest_Xray_Dataset/datasets/nih-chest-xrays/data/versions/3"
```

```python
df_train = pd.read_csv(os.path.join(data_dir, "df_train.csv"))
```

```python
df_train['Image Path'] = df_train['Image Path'].replace(to_replace="/usr/app/data", value= data_dir, regex=True)
```

```python
dataset = NIHChestXrayDataset(df_train, transform=None)
```

```python
train_dl = DataLoader(dataset, batch_size=32, shuffle=True)
```

```python
for batch in train_dl:
    try:
        images, labels = batch
        print("Images shape:", images.shape)
        print("Labels shape:", labels.shape)
    except RuntimeWarning as w:
        print("Warning processing batch:", w)
        break
```
