```python
import pandas as pd
import os
```

```python
data_folder = "/mnt/n4i_ds423/Employee_Folder/Jalil_Ahmed/NIH_Chest_Xray_Dataset/datasets/nih-chest-xrays/data/versions/3/"
```

```python
# Load metadata
df = pd.read_csv(os.path.join(data_folder, "Data_Entry_2017.csv"))

# Basic counts
n_images = len(df)
n_patients = df["Patient ID"].nunique()

# Demographics
age_mean = df["Patient Age"].mean()
age_std = df["Patient Age"].std()
sex_counts = df["Patient Gender"].value_counts()

# Disease labels
labels = [
    "Atelectasis", "Cardiomegaly", "Effusion", "Infiltration",
    "Mass", "Nodule", "Pneumonia", "Pneumothorax",
    "Consolidation", "Edema", "Emphysema", "Fibrosis",
    "Pleural_Thickening", "Hernia"
]

# Multi-label expansion
label_df = df["Finding Labels"].str.get_dummies(sep="|")

# Prevalence table
prevalence = (
    label_df[labels]
    .sum()
    .reset_index()
    .rename(columns={"index": "Condition", 0: "Count"})
)
prevalence["Percentage"] = 100 * prevalence["Count"] / n_images

# Assemble Table-1
table1 = {
    "Images": n_images,
    "Patients": n_patients,
    "Age (mean ± std)": f"{age_mean:.1f} ± {age_std:.1f}",
    "Male (%)": 100 * sex_counts.get("M", 0) / n_images,
    "Female (%)": 100 * sex_counts.get("F", 0) / n_images,
}

table1_df = pd.DataFrame.from_dict(table1, orient="index", columns=["Value"])

# Save outputs
table1_df.to_csv("../data/table1_dataset_summary.csv")
prevalence.to_csv("../data/table1_label_prevalence.csv", index=False)
```
